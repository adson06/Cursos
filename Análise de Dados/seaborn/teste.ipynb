```python
import seaborn as sns
import matplotlib as plt
import pandas as pd
```

```python
tabela = pd.read_excel('videosYT.xlsx')
display(tabela)
```

```python
tab_inscritos = pd.read_excel('videosYT.xlsx', 'Inscritos')
display(tab_inscritos)

graf_linha = sns.lineplot(data=tab_inscritos, x='Mês/Ano', y='Inscritos', color='blue')
graf_linha.set_title('Crescimento de Inscritos por Ano')
plt.show()
```

```python
fig = sns.displot(data = tabela,  x = "Nº de Likes", hue="Responsável", col="Categoria" )#kind = "kde", displot = gráfico de distribuição
plt.show(fig)
```
