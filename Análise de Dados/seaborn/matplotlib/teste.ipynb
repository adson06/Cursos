```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.plot(x, y) #line significa linha, então plt.plot é o gráfico de linha
plt.show() #mostra o gráfico
```

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]

plt.bar(x, y, color = 'green') #bar significa barra, então plt.bar é o gráfico de barras
plt.show() #mostra o gráfico
```

```python
plt.plot(x, y)
plt.axis('square') # define a proporção dos eixos x e y
plt.show() #mostra o gráfico
```

```python
x = [1, 2, 3, 4]
y =[1, 4, 3, 2]

figura = plt.figure(figsize=(20, 3)) #cria uma figura com tamanho personalizado
figura.suptitle('Título Geral') #adiciona um título à figura

figura.add_subplot(131) #adiciona um subplot à figura, 1 linha, 3 colunas, 1º subplot
plt.plot(x, y, label='Um dado qualquer') #adiciona um gráfico de linha ao subplot
plt.legend() #adiciona a legenda ao gráfico
plt.title('Gráfico1')

figura.add_subplot(132) #adiciona um subplot à figura, 1 linha, 3 colunas, 2º subplot
plt.bar(x, y, color='green', label='Outro dado qualquer') #adiciona um gráfico de barras ao subplot
plt.legend() #adiciona a legenda ao gráfico     
plt.title('Gráfico2')

figura.add_subplot(133) #adiciona um subplot à figura, 1 linha, 3 colunas, 3º subplot
plt.scatter(x, y, color='red', label='Outro dado qualquer') #adiciona um gráfico de dispersão ao subplot
plt.title('Gráfico3')
```

```python
import pandas as pd

cotacao1 = pd.read_csv('indexdata.csv')
display(cotacao1)
```

```python
cotacao_nya = cotacao1[['Index', 'Date', 'Close']] 
display(cotacao_nya)
```

```python
cotremove = cotacao_nya.loc[(cotacao_nya['Index'] != 'NYA')]
cotacao_final = cotacao_nya.drop(cotremove.index)
display(cotacao_final)
cotacao_final = cotacao_final[13900:]
```

```python

plt.figure(figsize=(20, 4)) 
plt.plot(cotacao_final['Date'], cotacao_final['Close'], label='Fechamento NYA', color='green', ls = '--', lw=2)
plt.legend()
plt.ylabel('Índice de Fechamento')
plt.xlabel('Data')
plt.title('Índice de Fechamento do NYA ao Longo do Tempo')
plt.xticks(['2021-03-23', '2021-04-23', '2021-05-03'])
plt.show()
```
